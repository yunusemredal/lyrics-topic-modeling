```python
import pandas as pd
import spotipy
from spotipy.oauth2 import SpotifyClientCredentials
from bs4 import BeautifulSoup
import requests
import json
cid = os.environ['SPOTIPY_CLIENT_ID']
secret = os.environ['SPOTIPY_CLIENT_SECRET']

client_credentials_manager = SpotifyClientCredentials(client_id=cid, client_secret=secret)
sp = spotipy.Spotify(client_credentials_manager = client_credentials_manager)
artiste = 'eminem'
def get_album_tracks(uri_info):
    uri = []
    track = []
    artist_name=[]
    duration = []
    explicit = []
    track_number = []
    popularity=[]
    one = sp.album_tracks(uri_info, limit=50, offset=0, market='US')
    df1 = pd.DataFrame(one)
    with open("example.json","w", encoding='utf-8') as jsonfile:
        json.dump(one,jsonfile,indent=2,ensure_ascii=False)



    for i, x in df1['items'].items():
        uri.append(x['uri'])
        track.append(x['name'])
        artist_name.append(x['artists'][0]['name'])
        duration.append(x['duration_ms'])
        explicit.append(x['explicit'])
        track_number.append(x['track_number'])

    df2 = pd.DataFrame({
    'uri':uri,
    'track':track,
    'artist_name' : artist_name,
    'duration_ms':duration,
    'explicit':explicit,
    'track_number':track_number})
    df2.to_csv(str(artiste)+'2.csv')

    return df2
def get_track_info(df):
    danceability = []
    energy = []
    key = []
    loudness = []
    speechiness = []
    acousticness = []
    instrumentalness = []
    liveness = []
    valence = []
    tempo = []
    for i in df['uri']:
        for x in sp.audio_features(tracks=[i]):
            danceability.append(x['danceability'])
            energy.append(x['energy'])
            key.append(x['key'])
            loudness.append(x['loudness'])
            speechiness.append(x['speechiness'])
            acousticness.append(x['acousticness'])
            instrumentalness.append(x['instrumentalness'])
            liveness.append(x['liveness'])
            valence.append(x['valence'])
            tempo.append(x['tempo'])

    df2 = pd.DataFrame({
    'danceability':danceability,
    'energy':energy,
    'key':key,
    'loudness':loudness,
    'speechiness':speechiness,
    'acousticness':acousticness,
    'instrumentalness':instrumentalness,
    'liveness':liveness,
    'valence':valence,
    'tempo':tempo})
    df2.to_csv(artiste+'.csv')
    return df2

def merge_frames(df1, df2):
    df3 = df1.merge(df2, left_index= True, right_index= True)
    return df3
Tr2Eng = str.maketrans("çğıöşüÇĞİÖŞÜ", "cgiosuCGIOSU")
def scrape_lyrics(artistname, songname):
    artist_name_eng= artistname.translate(Tr2Eng)
    songname_eng = songname.translate(Tr2Eng)
    artistname2 = str(artist_name_eng.replace(' ','-')) if ' ' in artist_name_eng else str(artist_name_eng)
    songname2 = str(songname_eng.replace(' ','-')) if ' ' in songname_eng else str(songname_eng)
    url = 'https://genius.com/'+ artistname2 + '-' + songname2 + '-' + 'lyrics'
    soup = BeautifulSoup(requests.get(url).content, 'lxml')
    dizis = []
    for tag in soup.select('div[class^="Lyrics__Container"], .song_body-lyrics p'):

        t = tag.get_text(strip=True, separator='\n')
        dizis.append(t)

        #with open(str(artistname2)+str(songname2)+".json","w", encoding='utf-8') as jsonfile:
         #   json.dump(dizis,jsonfile,indent=4,ensure_ascii=False)

        f = open(str(artistname2)+str(songname2)+".txt", "w")
        f.write(t)
    return

def lyrics_onto_frame(df1, artist_name):
    for i,x in enumerate(df1['track']):
        test = scrape_lyrics(artist_name, x)
        df1.loc[i, 'lyrics'] = test
    return df1



blonde_df1_tracks = get_album_tracks('spotify:album:604LjoILfRCmhYuLCFvel7')
blonde_df2_metadata = get_track_info(blonde_df1_tracks)
df1 = merge_frames(blonde_df1_tracks, blonde_df2_metadata)

lyrics_onto_frame(df1,'UZI')
```
