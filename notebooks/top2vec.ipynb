```python
import pandas as pd
import numpy as np
import re
import string
import time   
from top2vec import Top2Vec

```

```python
def top2vecDef(add_csv,add_index):
    query_df=pd.read_csv(add_csv)
    query_df = query_df.iloc[:8000]#subset of dataframe
    query_df[add_index] = query_df[add_index].astype(str)
    query_df[add_index].head()
    global docs
    docs = list(query_df.loc[:, add_index].values)
    topic_model = Top2Vec(
        docs, 
        embedding_model = 'doc2vec', # Embedding model: See [1,2] for supported models
        min_count = 30,              # Ignore words less frequent than this value
        umap_args = None,            # Dict of custom args for UMAP
        hdbscan_args = None          # Dict of custom argd for HDBSCAN
        )
    return topic_model
```

İngiltere 2016 ve 2019 kadar olan verileri Modellemeye attım 

```python
#İngiltere 2016 ve 2019 kadar olan verileri 
topic_model = top2vecDef('.csv','lyrics')
topic_model.get_num_topics()
```

```python
#İngiltere 2016 ve 2019 kadar olan verileri 
topic_words, word_scores, topic_scores, topic_nums = topic_model.search_topics(
    keywords = ["live"], 
    num_topics = 2)

# Plot the resulting topics as wordclouds
for topic in topic_nums:
    topic_model.generate_topic_wordcloud(topic)
```

```python
#İngiltere 2016 ve 2019 kadar olan verileri 
topic_words, word_scores, topic_nums = topic_model.get_topics(2)
for topic in topic_nums:
    topic_model.generate_topic_wordcloud(topic)
```

```python
for words, scores,num in zip(topic_words, word_scores, topic_nums):
    print(words, scores,num)
    print(f"words:{words}, scores:{scores}, num:{num}")
```
