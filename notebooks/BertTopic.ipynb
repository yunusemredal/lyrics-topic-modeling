```python
import pandas as pd 
from bertopic import BERTopic
from sentence_transformers import SentenceTransformer
from hdbscan import HDBSCAN
```

```python
def bertTopics(add_csv,add_index):
    query_df=pd.read_csv(add_csv)
    query_df = query_df.iloc[:8000]#subset of dataframe
    query_df[add_index] = query_df[add_index].astype(str)
    query_df[add_index].head()
    global docs
    docs = list(query_df.loc[:, add_index].values)
    embedding_model = SentenceTransformer("all-MiniLM-L6-v2")

    cluster_model = HDBSCAN(min_cluster_size = 15, 
                        metric = 'euclidean', 
                        cluster_selection_method = 'eom', 
                        prediction_data = True)
    topic_model = BERTopic(embedding_model = embedding_model,
                        hdbscan_model = cluster_model)

    topics, probs = topic_model.fit_transform(docs)
    return topic_model

```

İngiltere 2016 ve 2019 kadar olan verileri Modellemeye attım 


```python
#İngiltere 2016 ve 2019 kadar olan verileri 
topic_model = bertTopics('')
topic_model.get_topic_info()
```

```python
#İngiltere 2016 ve 2019 kadar olan verileri 
try:
    topic_model.visualize_topics()
except:
    print("Yeterli Veri Yok")
    pass
```

```python
#İngiltere 2016 ve 2019 kadar olan verileri 
topic_model.visualize_barchart(top_n_topics = 25)
```

```python
#İngiltere 2016 ve 2019 kadar olan verileri 
topic_model.visualize_hierarchy()
```

```python
#İngiltere 2016 ve 2019 kadar olan verileri 
topic_model.visualize_documents(docs)
```
